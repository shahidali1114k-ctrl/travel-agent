```python
# ==========================================
# LLM-Based Travel Agent (Gemini API)
# ==========================================

!pip install -U google-genai

from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"


def travel_agent(source, destination, budget):

    prompt = f"""
You are an AI Travel Agent Assistant that helps users plan a trip between two cities
based on their budget. You are NOT a booking agent — just give practical guidance
and suggestions.

Provide a well-structured response with clear headings:

1. **Best Travel Option**
   - Recommend the most suitable mode of transport (flight, bus, train, etc.)
     for this budget and route
   - Briefly explain why this is the best choice

2. **Estimated Cost Breakdown**
   - Approximate cost of transport
   - Approximate cost of food/snacks during travel
   - Any other possible expenses (in case budget allows)

3. **Estimated Travel Time**
   - How long the journey will take with the recommended option

4. **Money-Saving Tips**
   - 2-3 practical tips to stay within budget

5. **Recommended Next Step**
   - What the user should do next (e.g., book tickets in advance, check bus terminal timing, etc.)

Keep the tone friendly, clear, and easy to understand for a non-expert user.
Always end with this disclaimer on its own line:
"⚠️ This is general travel guidance only. Please confirm current prices and schedules before booking."

User's Trip Details:
- From: {source}
- To: {destination}
- Budget: Rs. {budget}

Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


def main():
    print("Gemini LLM-Based Travel Agent Started (type 'exit' to stop)\n")

    while True:
        source = input("Apna departure city likhen (ya 'exit' likhen): ").strip()

        if source.lower() == "exit":
            print("Goodbye 👋")
            break

        destination = input("Apna destination city likhen: ").strip()

        try:
            budget = int(input("Apna budget likhen (Rs.): "))
        except ValueError:
            print("Invalid budget! Sirf number likhen.")
            continue

        try:
            reply = travel_agent(source, destination, budget)
            print("\nAgent:\n", reply, "\n")
        except Exception as e:
            print("Error:", e)


if __name__ == "__main__":
    main()
```
