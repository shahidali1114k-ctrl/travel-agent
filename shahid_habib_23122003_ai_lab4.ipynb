```python
!pip install -U google-genai

from google import genai
client = genai.Client(api_key="")

MODEL = "models/gemini-3.6-flash"

# ==========================================
# FUUAST Library Information Database
# ==========================================
fuuast_library_info = {
    "university": {
        "name": "Federal Urdu University of Arts, Science & Technology",
        "short_name": "FUUAST",
        "campus": "Gulshan-e-Iqbal, Karachi",
        "website": "https://www.fuuast.edu.pk"
    },

    "library": {
        "name": "FUUAST Central Library",
        "location": "Main Campus, Gulshan-e-Iqbal, Karachi",
        "in_charge": "Chief Librarian (Library Department)",

        "timing": {
            "Monday to Friday": "9:00 AM - 4:00 PM",
            "Saturday": "Closed",
            "Sunday": "Closed",
            "Ramzan Timing": "9:00 AM - 1:00 PM (Adjusted)"
        },

        "membership": {
            "eligibility": "All enrolled FUUAST students and faculty",
            "documents_required": ["Student ID Card", "Admission Slip", "Passport size photo"],
            "fee": "One-time Rs. 200 security (refundable)"
        },

        "book_issue_rules": {
            "students": {
                "books_allowed": 2,
                "duration": "14 days",
                "renewal": "1 time, if not reserved by others"
            },
            "faculty": {
                "books_allowed": 5,
                "duration": "30 days",
                "renewal": "2 times"
            },
            "fine_per_day": "Rs. 5 per book after due date"
        },

        "sections": {
            "General Section": "Books on Arts, Social Sciences, Urdu Literature",
            "Science Section": "Computer Science, Mathematics, Physics, Chemistry",
            "Reference Section": "Encyclopedias, Dictionaries (Not issued, reading only)",
            "Thesis Section": "MPhil/PhD theses and research papers",
            "Periodicals Section": "Newspapers, Magazines, Journals"
        },

        "computer_science_books": {
            "AI": "Artificial Intelligence by Stuart Russell (Science Section)",
            "Programming": "Python/C++ Programming books (Science Section)",
            "Networks": "Computer Networks by Tanenbaum (Science Section)",
            "Databases": "Database Systems by Elmasri (Science Section)"
        },

        "facilities": {
            "reading_hall": "Available, 100+ seating capacity",
            "wifi": "Available for students",
            "photocopy": "Available at nominal charges",
            "digital_library": "HEC Digital Library access for research"
        }
    },

    "announcements": {
        "new_arrivals": {
            "title": "New Books Arrived",
            "description": "Latest editions of Computer Science and AI books have been added to the Science Section."
        },
        "fine_waiver": {
            "title": "Fine Waiver Notice",
            "description": "Students can request fine waiver once per semester with valid reason, approved by librarian."
        }
    }
}


def fuuast_library_agent(user_input):
    prompt = f"""
You are FUUAST Library Assistant, an AI agent for the Federal Urdu University of Arts,
Science & Technology (FUUAST) Central Library.

Use this information as your main source when relevant:

LIBRARY_INFO:
{fuuast_library_info}

Answer questions about library timing, book issue rules, fine, sections, membership,
facilities and computer science books using this information.

For anything else, answer normally using your own knowledge.

Keep answers short, clear and simple.

Question:
{user_input}

Answer:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST Library Assistant Agent")
print("Federal Urdu University - Central Library")
print("Type 'exit' to stop\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye!")
        break

    try:
        reply = fuuast_library_agent(user)
        print("\nAgent:", reply)
        print()

    except Exception as e:
        print("Error:", e)
```
