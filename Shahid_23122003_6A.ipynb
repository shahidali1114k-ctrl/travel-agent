```python
# ==========================================
# Travel Agent System (Multi-City)
# ==========================================

# Rule database: (source, destination) -> list of options based on budget
travel_rules = {
    ("karachi", "islamabad"): [
        (15000, "✈️ Flight - Approx Rs. 12,000-18,000 | Time: 2 hours"),
        (8000,  "🚌 AC Business Bus (Daewoo) - Approx Rs. 6,000-8,000 | Time: 18-20 hrs"),
        (5000,  "🚌 Standard AC Bus - Approx Rs. 3,500-5,000 | Time: 20-22 hrs"),
        (0,     "🚆 Train (Economy) - Approx Rs. 2,000-3,000 | Time: 24+ hrs")
    ],
    ("karachi", "lahore"): [
        (14000, "✈️ Flight - Approx Rs. 10,000-14,000 | Time: 1.5 hours"),
        (7000,  "🚌 AC Business Bus - Approx Rs. 5,000-7,000 | Time: 16-18 hrs"),
        (4500,  "🚌 Standard AC Bus - Approx Rs. 3,000-4,500 | Time: 18-20 hrs"),
        (0,     "🚆 Train (Economy) - Approx Rs. 1,800-2,500 | Time: 20+ hrs")
    ],
    ("karachi", "hyderabad"): [
        (2000,  "🚌 AC Coaster - Approx Rs. 1,200-2,000 | Time: 2-3 hrs"),
        (800,   "🚌 Local Bus - Approx Rs. 400-800 | Time: 3 hrs"),
        (0,     "🚆 Train (Economy) - Approx Rs. 300-500 | Time: 3-4 hrs")
    ],
    ("islamabad", "murree"): [
        (3000,  "🚗 Private Taxi/Rental Car - Approx Rs. 2,500-3,500 | Time: 1.5 hrs"),
        (1000,  "🚌 Local Wagon/Van - Approx Rs. 500-1,000 | Time: 2 hrs"),
        (0,     "🚌 Shared Van (budget) - Approx Rs. 300-500 | Time: 2-2.5 hrs")
    ],
    ("lahore", "murree"): [
        (5000,  "🚗 Private Taxi/Rental Car - Approx Rs. 4,000-5,000 | Time: 5-6 hrs"),
        (2500,  "🚌 AC Bus + Local Van combo - Approx Rs. 1,800-2,500 | Time: 7-8 hrs"),
        (0,     "🚌 Standard Bus (budget) - Approx Rs. 1,200-1,800 | Time: 8-9 hrs")
    ],
    ("karachi", "gwadar"): [
        (10000, "✈️ Flight - Approx Rs. 8,000-10,000 | Time: 1 hour"),
        (5000,  "🚌 AC Coach - Approx Rs. 3,500-5,000 | Time: 10-12 hrs"),
        (0,     "🚌 Standard Bus - Approx Rs. 2,000-3,500 | Time: 12-14 hrs")
    ],
    ("islamabad", "hunza"): [
        (18000, "✈️ Flight (to Gilgit) + Local Taxi - Approx Rs. 15,000-18,000 | Time: 2 hrs + 3 hrs"),
        (6000,  "🚌 NATCO Bus - Approx Rs. 4,000-6,000 | Time: 18-20 hrs"),
        (0,     "🚐 Shared Hiace Van (budget) - Approx Rs. 2,500-4,000 | Time: 20-22 hrs")
    ],
}

# List of villages/small towns that connect via nearby big cities
villages_info = {
    "murree": "Hill station near Islamabad/Lahore - popular for short trips",
    "hunza": "Village in Gilgit-Baltistan - scenic valley, accessible via Islamabad",
    "gwadar": "Coastal town in Balochistan - accessible via Karachi",
    "skardu": "Mountain town in Gilgit-Baltistan - accessible via Islamabad (flight recommended)",
}


def get_travel_plan(source, destination, budget):
    key = (source, destination)

    if key not in travel_rules:
        return "❌ Sorry, is route ke liye abhi data available nahi hai.\n" \
               "Available routes dekhne ke liye 'show_routes()' function use karen."

    options = travel_rules[key]

    # Rule-based selection: budget ke range ke mutabiq best option choose karna
    for min_budget, plan in options:
        if budget >= min_budget:
            return plan

    return "Budget bohat kam hai, koi option available nahi."


def show_routes():
    print("\n📍 Available Routes:")
    for (src, dest) in travel_rules.keys():
        print(f" - {src.title()} ➝ {dest.title()}")


def show_villages():
    print("\n🏞️ Villages/Small Towns Info:")
    for village, info in villages_info.items():
        print(f" - {village.title()}: {info}")


def main():
    print("=== Welcome to Rule-Based Travel Agent (Multi-City) ===")
    show_routes()
    show_villages()

    source = input("\nApna departure city likhen: ").strip().lower()
    destination = input("Apna destination city likhen: ").strip().lower()

    try:
        budget = int(input("Apna budget likhen (Rs.): "))
    except ValueError:
        print("Invalid budget! Sirf number likhen.")
        return

    print("\n--- Aapka Travel Plan ---")
    plan = get_travel_plan(source, destination, budget)
    print(plan)


if __name__ == "__main__":
    main()
```
